```python
acc['Number_of_Vehicles'].value_counts().sort_index().head(8).plot(kind='bar', color='green')
plt.title('Vehicles involved per accident')
plt.xlabel('Number of Vehicles')
plt.ylabel('Accidents')
plt.show()
```


- 2 vehicles: Most common (~1.2 million accidents).
- 1 vehicle: Second most common (~600,000 accidents).
- Higher vehicle counts decrease quickly.

Explanation: The bar chart shows how many accidents involved different numbers of vehicles. Most accidents involved 2 vehicles, followed by 1 vehicle.

```python
sns.relplot(data=acc, x='Number_of_Vehicles', y='Number_of_Casualties', kind='scatter')
```


- Most points are concentrated at low vehicle and casualty counts.
- A few outliers exist, with casualties reaching up to 93.

Explanation: The scatter plot shows the relationship between the number of vehicles and casualties in road accidents.

```python
sns.relplot(data=acc, x='Number_of_Vehicles', y='Number_of_Casualties',
            kind='scatter', hue='Weather_Conditions',
            style='Light_Conditions', size='Number_of_Casualties')
```


- Most accident records are associated with fine weather and daylight.
- Different colors represent weather conditions.
- Different styles represent light conditions.
- Point sizes represent casualty counts.

Explanation: This visualization helps compare road accidents under different weather and lighting conditions. Casualty numbers are also represented through point sizes.

```python
sns.relplot(data=acc, x='Number_of_Vehicles', y='Number_of_Casualties',
            kind='line')
```


- The line remains almost flat up to around 33 vehicles.
- Results beyond 30 vehicles are less reliable because very few accidents involve so many vehicles.

Explanation: The line plot shows the general relationship between vehicles and casualties. However, values at higher vehicle counts should be interpreted carefully.

```python
sns.heatmap(acc.corr(numeric_only=True), annot=True, cmap='YlGnBu')
plt.title('Road Accident Correlation')
plt.show()
```


- Most variables have weak correlations.
- The strongest meaningful relationship is Casualties vs Vehicles (0.24).
- Latitude vs Northing has a correlation of 0.999973.

Explanation: The heatmap represents the correlation between numerical variables using different colors. It helps identify strong and weak relationships in the dataset.
